## 2D Convex Hull

```python
num_points = 100
points = np.random.randn(num_points, 2)

convex_utils = importlib.reload(convex_utils)
convex_hull = convex_utils.convex_hull_2d(points)
```

```python
def plot_line_segment(start, end, **kwargs):
    x = [start[0], end[0]]
    y = [start[1], end[1]]
    return go.Scatter(x=x, y=y, mode="lines", **kwargs)

# Scatter plot of the points
# With labels as the index of the points, drawn on the plot
scatters = []
scatter_all = go.Scatter(
    x=points[:, 0],
    y=points[:, 1],
    mode="markers+text",
    marker=dict(size=5, color="blue"),
    text=[str(i) for i in range(num_points)],
    textposition="top center",
)
scatters.append(scatter_all)

for idx in range(len(convex_hull)):
    start_idx = convex_hull[idx]
    end_idx = convex_hull[(idx + 1) % len(convex_hull)]
    start = points[start_idx, :]
    end = points[end_idx, :]
    scatters.append(plot_line_segment(
        start, end, **{"line": {"color": "rgba(255, 0, 0, 0.5)"}, "showlegend": False}
    ))

fig = go.Figure(data=scatters)
fig.update_layout(
    title="Scatter plot of random points and their convex hull",
    width=800,
    height=800,
)
fig.show()
```
